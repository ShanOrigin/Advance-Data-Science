## **1. Analyze Customer Purchase Patterns**

Given a CSV file containing customer transactions (`customer_id`, `purchase_amount`, `date`), use Pandas to calculate the average monthly spending per customer.

```python
import pandas as pd
import numpy as np

# Creating some realistic-looking demo data for customer transactions
np.random.seed(42) # For reproducibility

num_customers = 5
customer_ids = [f'C_{i:03d}' for i in range(1, num_customers + 1)]

# Generate dates over a few months
dates = pd.to_datetime(pd.date_range(start='2023-01-01', end='2023-03-31', freq='D').tolist() * 2)

# Create a list to hold transaction records
transaction_records = []
for _ in range(200): # Simulate 200 transactions
    customer_id = np.random.choice(customer_ids)
    purchase_amount = round(np.random.uniform(10, 500), 2) # Random purchase amounts
    transaction_date = np.random.choice(dates)
    transaction_records.append([customer_id, purchase_amount, transaction_date])

# Convert to a DataFrame
customer_transactions_df = pd.DataFrame(transaction_records, columns=['customer_id', 'purchase_amount', 'date'])

# Display the first few rows of the generated data
print("Demo Customer Transaction Data:")
display(customer_transactions_df.head())
```

```python
# Ensure the 'date' column is in datetime format
customer_transactions_df['date'] = pd.to_datetime(customer_transactions_df['date'])

# Extract the month and year to group by
customer_transactions_df['year_month'] = customer_transactions_df['date'].dt.to_period('M')

# Group by customer_id and year_month, then calculate the sum of purchase_amount
monthly_spending = customer_transactions_df.groupby(['customer_id', 'year_month'])['purchase_amount'].sum().reset_index()

# Calculate the average monthly spending per customer
average_monthly_spending_per_customer = monthly_spending.groupby('customer_id')['purchase_amount'].mean().reset_index()

average_monthly_spending_per_customer.rename(columns={'purchase_amount': 'average_monthly_spending'}, inplace=True)

print("Average Monthly Spending Per Customer:")
display(average_monthly_spending_per_customer)
```

## **2. Identify Missing Values and Impute**

Using a dataset with missing values (e.g., `age`, `salary`), use Pandas and NumPy to identify missing values and fill them with the column mean.

```python
import pandas as pd
import numpy as np

# Creating some demo data with missing values
data_with_missing = {
    'employee_id': range(1, 11),
    'age': [25, 30, np.nan, 35, 40, np.nan, 28, 32, 45, 38],
    'salary': [50000, 60000, 75000, np.nan, 80000, 55000, 70000, np.nan, 90000, 62000],
    'department': ['HR', 'IT', 'Marketing', 'IT', 'HR', 'Sales', 'IT', 'Marketing', 'Sales', 'HR']
}
missing_data_df = pd.DataFrame(data_with_missing)

print("Demo Data with Missing Values:")
display(missing_data_df)
```

```python
# Identify missing values
print("Missing values before imputation:")
display(missing_data_df.isnull().sum())

# Impute missing values in 'age' column with its mean
mean_age = missing_data_df['age'].mean()
missing_data_df['age'].fillna(mean_age, inplace=True)

# Impute missing values in 'salary' column with its mean
mean_salary = missing_data_df['salary'].mean()
missing_data_df['salary'].fillna(mean_salary, inplace=True)

print("\nMissing values after imputation:")
display(missing_data_df.isnull().sum())

print("\nDataFrame after imputing missing values:")
display(missing_data_df)
```

## **3. Filter and Sort Employee Records**

Given employee data (`name`, `department`, `salary`), use Pandas to filter employees in the `"IT"` department earning above 60K and sort them by salary.

```python
import pandas as pd

# Creating demo employee data
employee_data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'department': ['HR', 'IT', 'Sales', 'IT', 'Marketing', 'IT', 'HR', 'IT'],
    'salary': [55000, 65000, 70000, 72000, 60000, 58000, 80000, 78000]
}
employees_df = pd.DataFrame(employee_data)

print("Demo Employee Data:")
display(employees_df)
```

```python
# Filter for employees in the 'IT' department
it_employees = employees_df[employees_df['department'] == 'IT']

# Further filter for those earning above 60K
high_earning_it_employees = it_employees[it_employees['salary'] > 60000]

# Sort the filtered employees by salary in ascending order
sorted_it_employees = high_earning_it_employees.sort_values(by='salary', ascending=True)

print("IT Department Employees Earning Above 60K, Sorted by Salary:")
display(sorted_it_employees)
```

## **4. Calculate Statistical Metrics for Stock Data**

Using NumPy and Pandas, calculate the mean, median, and standard deviation of daily stock prices from a dataset.

```python
import pandas as pd
import numpy as np

# Generating demo stock price data
np.random.seed(42) # for reproducibility
dates = pd.date_range(start='2023-01-01', periods=30, freq='D')
daily_prices = np.random.normal(loc=150, scale=10, size=len(dates)).round(2)

stock_data = pd.DataFrame({
    'date': dates,
    'price': daily_prices
})

print("Demo Daily Stock Prices:")
display(stock_data.head())
```

```python
# Calculate the mean of daily stock prices
mean_price = stock_data['price'].mean()

# Calculate the median of daily stock prices
median_price = stock_data['price'].median()

# Calculate the standard deviation of daily stock prices
std_dev_price = stock_data['price'].std()

print(f"Mean Daily Stock Price: ${mean_price:.2f}")
print(f"Median Daily Stock Price: ${median_price:.2f}")
print(f"Standard Deviation of Daily Stock Prices: ${std_dev_price:.2f}")
```

## **5. Create a Pivot Table for Sales Analysis**

From a sales dataset (`region`, `product`, `sales`), use Pandas to create a pivot table showing total sales per region per product.

```python
import pandas as pd
import numpy as np

# Generating demo sales data
np.random.seed(42)

regions = ['North', 'South', 'East', 'West']
products = ['Laptop', 'Mouse', 'Keyboard', 'Monitor']

sales_records = []
for _ in range(100): # 100 sales transactions
    region = np.random.choice(regions)
    product = np.random.choice(products)
    sales = round(np.random.uniform(50, 1500), 2) # Random sales amount
    sales_records.append({'region': region, 'product': product, 'sales': sales})

sales_df = pd.DataFrame(sales_records)

print("Demo Sales Data:")
display(sales_df.head())
```

```python
# Create a pivot table to show total sales per region per product
sales_pivot_table = pd.pivot_table(
    sales_df,
    values='sales',
    index='region',
    columns='product',
    aggfunc='sum',
    fill_value=0 # Fill any NaN values (where a product wasn't sold in a region) with 0
)

print("Total Sales Per Region Per Product (Pivot Table):")
display(sales_pivot_table)
```

## **6. Plot Monthly Sales Using Matplotlib**

From a dataset with `month` and `sales`, use Matplotlib to create a line chart showing monthly sales trends.

```python
import pandas as pd
import numpy as np

# Generating demo monthly sales data for a year
np.random.seed(42)

months = pd.date_range(start='2023-01-01', periods=12, freq='M').strftime('%Y-%m')
monthly_sales_values = np.random.randint(10000, 50000, size=len(months))

monthly_sales_df = pd.DataFrame({
    'month': months,
    'sales': monthly_sales_values
})

print("Demo Monthly Sales Data:")
display(monthly_sales_df)
```

```python
import matplotlib.pyplot as plt

# Ensure 'month' is treated as a categorical or ordered type for plotting if it's not a datetime object
# Here, it's already strings like 'YYYY-MM', which works well for x-axis labels

plt.figure(figsize=(10, 6))
plt.plot(monthly_sales_df['month'], monthly_sales_df['sales'], marker='o', linestyle='-', color='skyblue')

plt.title('Monthly Sales Trend (2023)', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45) # Rotate month labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

## **7. Compare Category Sales with Bar Chart**

Create a bar chart using Matplotlib to compare total sales for different product categories.

```python
import pandas as pd
import numpy as np

# Generating demo sales data by category
np.random.seed(42)

categories = ['Electronics', 'Clothing', 'Home Goods', 'Books', 'Groceries']

category_sales_records = []
for _ in range(200): # Simulate 200 sales transactions
    category = np.random.choice(categories)
    sale_amount = round(np.random.uniform(20, 500), 2)
    category_sales_records.append({'category': category, 'sale_amount': sale_amount})

category_sales_df = pd.DataFrame(category_sales_records)

# Aggregate total sales per category
total_sales_by_category = category_sales_df.groupby('category')['sale_amount'].sum().reset_index()
total_sales_by_category = total_sales_by_category.sort_values(by='sale_amount', ascending=False)

print("Demo Aggregated Category Sales Data:")
display(total_sales_by_category)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(total_sales_by_category['category'], total_sales_by_category['sale_amount'], color='lightcoral')

plt.title('Total Sales by Product Category', fontsize=16)
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Sales Amount', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotate category labels if they overlap
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add horizontal grid lines
plt.tight_layout() # Adjust layout
plt.show()
```

## **8. Visualize Correlation Using Heatmap (Seaborn)**

Given a dataset with multiple numeric columns, use Seaborn to draw a heatmap of the correlation matrix.

```python
import pandas as pd
import numpy as np

# Generating demo dataset with multiple numeric columns
np.random.seed(42)

data_corr = {
    'feature_A': np.random.rand(50) * 100,
    'feature_B': np.random.rand(50) * 50 + (np.random.rand(50) * 20), # Some positive correlation
    'feature_C': np.random.rand(50) * 200 - (np.random.rand(50) * 30), # Some negative correlation
    'feature_D': np.random.rand(50) * 75
}
correlation_df = pd.DataFrame(data_corr)

print("Demo Numeric Data for Correlation:")
display(correlation_df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the correlation matrix
correlation_matrix = correlation_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True, # Show the correlation values on the heatmap
    cmap='coolwarm', # Color map for the heatmap
    fmt=".2f", # Format annotation to two decimal places
    linewidths=.5 # Add lines between cells
)
plt.title('Correlation Matrix Heatmap', fontsize=16)
plt.show()
```

## **9. Visualize Age Distribution Using Histogram**

Using a dataset with an `age` column, plot a histogram with Seaborn to show the age distribution.

```python
import pandas as pd
import numpy as np

# Generating demo age data
np.random.seed(42)

# Simulate ages, centered around 35 with some spread, for 100 individuals
ages = np.random.normal(loc=35, scale=10, size=100)
ages = np.clip(ages, 18, 70).astype(int) # Ensure ages are within a reasonable range

age_data_df = pd.DataFrame({'age': ages})

print("Demo Age Data:")
display(age_data_df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.histplot(
    age_data_df['age'],
    bins=10, # Number of bins for the histogram
    kde=True, # Overlay a kernel density estimate curve
    color='teal', # Color of the bars
    edgecolor='black' # Color of the bar edges
)

plt.title('Distribution of Age', fontsize=16)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of Individuals', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## **10. Scatter Plot to Show Relationship Between Two Variables**

Use Seaborn to create a scatter plot showing the relationship between advertising budget and sales.

```python
import pandas as pd
import numpy as np

# Generating demo data for advertising budget and sales
np.random.seed(42)

# Simulate advertising budget (e.g., in thousands of dollars)
advertising_budget = np.random.rand(50) * 100 + 10 # 10 to 110 thousand

# Simulate sales, with a positive correlation to budget, plus some noise
sales = 50 + (advertising_budget * 2.5) + np.random.normal(loc=0, scale=20, size=50) # Base sales + budget effect + noise
sales = np.clip(sales, 0, None) # Ensure sales are not negative

marketing_data_df = pd.DataFrame({
    'advertising_budget': advertising_budget,
    'sales': sales
})

print("Demo Advertising Budget and Sales Data:")
display(marketing_data_df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='advertising_budget',
    y='sales',
    data=marketing_data_df,
    hue='advertising_budget', # Color points based on budget for added visual insight
    size='sales', # Size points based on sales for added visual insight
    sizes=(50, 500), # Range of sizes for the points
    palette='viridis', # Color palette
    alpha=0.7 # Transparency of points
)

plt.title('Advertising Budget vs. Sales', fontsize=16)
plt.xlabel('Advertising Budget (in Thousands)', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
