### Question 2 - Central Tendency: Mean, Median and Mode

Write a Python **function** to compute the **mean**, **median**, and **mode** of a list of numbers.

**Requirements:**
- Define separate functions: `compute_mean(data)`, `compute_median(data)`, `compute_mode(data)`.
- Do **not** use the `statistics` module — implement the logic manually.
- Test your functions with at least two different datasets (one with a unique mode, one with multiple modes).
- Print the result for each measure clearly.

> _Formula reference:_
> - Mean: $\bar{x} = \dfrac{\sum x_i}{n}$
> - Median: Middle value of the sorted list (average of two middle values if $n$ is even)
> - Mode: Value(s) that appear most frequently

```python
# Question 2 — Mean, Median and Mode
# Write your code below

def mean(data):
    if not data:
        return None
    return sum(data) / len(data)

def median(data):
    if not data:
        return None
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2
    else:
        return sorted_data[mid]

def mode(data):
    if not data:
        return None

    frequency = {}
    for item in data:
        frequency[item] = frequency.get(item, 0) + 1

    max_frequency = 0
    if frequency:
        max_frequency = max(frequency.values())

    if max_frequency == 0:
        return None

    modes = [key for key, value in frequency.items() if value == max_frequency]
    return modes


data1 = [1, 2, 2, 3, 4, 5, 5, 5, 6, 7]
print("Data:", data1)
print(f"Mean   : {mean(data1)}")
print(f"Median : {median(data1)}")
print(f"Mode   : {mode(data1)}")

data2 = [10, 20, 30, 20, 10, 40, 50, 40]
print("\n")
print("Data:", data2)
print(f"Mean   : {mean(data2)}")
print(f"Median : {median(data2)}")
print(f"Mode   : {mode(data2)}")



# Expected Output:
# Mean   : <value>
# Median : <value>
# Mode   : <value(s)>
```

### Question 3 - Measures of Dispersion: Variance and Standard Deviation

Write a Python **function** to compute the **variance** and **standard deviation** of a list of numbers.

**Requirements:**
- Define functions: `compute_variance(data)` and `compute_std_dev(data)`.
- Implement both **population** and **sample** variants.
- Do **not** use the `statistics` or `numpy` module — implement the logic manually.
- Test your functions with at least two different datasets.

> _Formula reference:_
> - Population Variance: $\sigma^2 = \dfrac{\sum (x_i - \bar{x})^2}{n}$
> - Sample Variance: $s^2 = \dfrac{\sum (x_i - \bar{x})^2}{n - 1}$
> - Standard Deviation: $\sigma = \sqrt{\sigma^2}$

```python

def variance(data):
    if not data or len(data) == 0:
        return {'population_variance': None, 'sample_variance': None}

    n = len(data)
    m = mean(data)

    if m is None:
        return {'population_variance': None, 'sample_variance': None}

    sum_sq_diff = sum([(x - m) ** 2 for x in data])

    population_variance = sum_sq_diff / n

    sample_variance = None
    if n > 1:
        sample_variance = sum_sq_diff / (n - 1)

    return {
        'population_variance': population_variance,
        'sample_variance': sample_variance
    }
```

```python

def std_dev(variance_results):
    population_variance = variance_results.get('population_variance')
    sample_variance = variance_results.get('sample_variance')

    population_std_dev = None
    if population_variance is not None:
        population_std_dev = population_variance ** 0.5

    sample_std_dev = None
    if sample_variance is not None:
        sample_std_dev = sample_variance ** 0.5

    return {
        'population_std_dev': population_std_dev,
        'sample_std_dev': sample_std_dev
    }
```

```python

data1 = [60, 65, 70, 75, 80]
variance_data1 = variance(data1)
std_dev_data1 = std_dev(variance_data1)
```

```python

print(f"Population Variance : {variance_data1['population_variance']:.2f}")
print(f"Sample Variance     : {variance_data1['sample_variance']:.2f}")
print(f"Population Std Dev  : {std_dev_data1['population_std_dev']:.2f}")
print(f"Sample Std Dev      : {std_dev_data1['sample_std_dev']:.2f}")
```

### Question 4 - Frequency Distribution Table

| **(b)** | Determine the **number of classes** using Sturges formula: $k = 1 + 3.322 \log_{10} n$ |

```python
# Question 4(a) — Determine the Range of the raw data
# Write your code below

x_max = data.max()
x_min = data.min()
range_data = x_max - x_min
print(f"Range = {range_data}")

# Expected Output: Range = <value>
```

| **(c)** | Determine the **class width**: $h = \lceil \text{Range} / k \rceil$ |

```python
# Question 4(b) — Number of Classes (Sturges formula)
# Write your code below

n = len(data)
k = 1 + 3.322 * np.log10(n)
k = math.ceil(k)
print(f"Number of Classes (k) = {k}")

# Expected Output: Number of Classes (k) = <value>
```

| **(d)** | Determine the **midpoint** of each class interval: $m = (\text{lower} + \text{upper}) / 2$ |

```python
# Question 4(c) — Class Width
# Write your code below

h = math.ceil(range_data / k)
print(f"Class Width (h) = {h}")

# Expected Output: Class Width (h) = <value>
```

| **(e)** | Determine the **relative frequency** of each class: $f_{rel} = f / n$ |

```python
# Question 4(d) — Midpoint of each class interval
# Write your code below

class_intervals = []
lb = x_min
for _ in range(k):
    up = lb + h
    class_intervals.append((lb, up))
    lb = up

midpoints = [float(((interval[0] + interval[1]) / 2)) for interval in class_intervals]
print("List of class midpoints:", midpoints)

# Expected Output: List of class midpoints
```

| **(f)** | Determine the **cumulative frequency** of each class |

```python
# Question 4(e) — Relative Frequency
# Write your code below

class_labels = []
freqs = []
lb = x_min

for i in range(k):
    up = lb + h
    # Handle the last interval to include max value if it falls exactly on the upper bound
    if i == k - 1:
        class_label = f"[{lb:.0f} - {up:.0f}]"
        count = np.sum((data >= lb) & (data <= up))
    else:
        class_label = f"[{lb:.0f} - {up:.0f})"
        count = np.sum((data >= lb) & (data < up))

    class_labels.append(class_label)
    freqs.append(count)
    lb = up

relative_frequencies = [float(freq / n) for freq in freqs]
print("relative freqs:", relative_frequencies)

# Expected Output: List of relative freqs (should sum to 1.0)
```
